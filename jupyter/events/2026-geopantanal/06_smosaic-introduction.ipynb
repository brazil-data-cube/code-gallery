# Introduction
<hr style="border:1px solid #0077b9;">

The **smosaic** package is a free, open-source Python tool designed to create Analysis-Ready Data (ARD) seamless, cloud-free mosaics from remote sensing imagery. Built to be accessible for analysts without deep programming knowledge, it allows you to retrieve a spatially contiguous composite as a Cloud-Optimized GeoTIFF (COG) simply by providing a location and a time interval.

A core strength of smosaic is its native integration with the SpatioTemporal Asset Catalog (STAC), a standard officially recognized by the Open Geospatial Consortium (OGC). This interoperability allows the package to seamlessly connect to and retrieve multi-sensor data from several major catalogs, including the **Brazil Data Cube (BDC)**, **Digital Earth Africa**, **Swiss Data Cube**, and the **Microsoft Planetary Computer**.

In smosaic, a mosaic is generated by merging the best available pixels from a collection of satellite scenes. The package provides different composition methods, allowing users to generate mosaics using `lcf` (Least Cloud-Cover First), `ctd` (Closest to Date), or `chrono` (Chronological) compose functions. 

To ensure scalability when working with large-scale environmental monitoring, smosaic leverages parallel computing via Python's `multiprocessing` library, distributing temporal clipping and ordering operations to run concurrently.

This Jupyter Notebook demonstrates how to use smosaic to build ARD composites.

## Operations

The workflow is orchestrated through the following functions:

- `mosaic()`: Orchestrates the entire workflow. It establishes the spatiotemporal composition schema based on user-defined parameters.
  
- `collection_get_data()`: Interfaces with a STAC client to query, download, and locally organize all the available satellite images by scene for the given region.
  
- `process_period()`: Opens, filters, and clips the downloaded images in parallel based on the chosen composition function (`lcf`, `ctd`, or `chrono`).
  
- `merge_tifs()`: Fuses the processed data into a final Cloud-Optimized GeoTIFF (COG), complying with OGC standards for highly efficient web-based geospatial processing.

Along with the user’s requested spectral bands, the package also provides the cloud band and a provenance band—a layer that indicates the origin date of the image selected as the best pixel for the composition.

# Generating a image mosaic
<hr style="border:1px solid #0077b9;">

In order to generate a image mosaic `red`, `green` and `blue`, in the BDC tile_id `020019` from `January 1st, 2026` to `January 16st, 2026`, use the `mosaic()` function with the `grid`, `tile_id` and `projection_output` parameters:

````python
result = mosaic(
    name="MT",
    data_dir=os.path.abspath(""),
    stac_url=stac_url,
    collection="S2_L2A-1",
    grid="BDC_SM_V2",
    tile_id="020019",
    projection_output="BDC",
    output_dir=os.path.join("output"),   
    mosaic_method="lcf", 
    start_year=2026,
    start_month=1,
    start_day=1,
    duration_days=16, 
    bands=["B02","B03","B04"]
)
````

You can generate a mosaic using a custom geographic area instead of the standard BDC tile grid. Simply replace the tile parameters (`grid` and `tile_id`) with a single `bbox` parameter that defines your area of interest.
